### Question 8

The `ggplot2` dataset `diamonds` contains the price and information from the sale of 54,000 different diamonds. You can use ?diamonds to find out more about the dataset if you want.

Using `ggplot2`:

- plot the relationship between the carat (i.e. the weight of the diamond) and the price.
- plot a histogram of prices of all the diamonds
- plot a boxplot of the difference in price between the different cuts

```r
install.packages("ggplot2")
library(ggplot2)
data(diamonds) #Loads data set
ggplot(aes(carat, price), data = diamonds)+ #Makes axis
    geom_point(alpha = 0.3, colour = "navy") #Plots as scatter graph

ggplot(aes(price), data = diamonds)+ #Makes axis
    geom_histogram(binwidth = 500) #Plots as histogram

ggplot(aes(cut, price), data = diamonds)+ #Makes axis
    geom_boxplot() #Plots as boxplot
```

### Question 9

Identify the maximum and minimum price that the diamonds in the diamonds data set sold for. Add a new column to the diamonds data set that contains the string “Low price” if the value in the price column is less than 5000 dollars and “High price” if it is above 5000 dollars.

```r
data(diamonds) #Loads data set
head(arrange(diamonds, desc(price)), 1) #Finds most expensive
head(arrange(diamonds, price), 1) #Finds cheapest

diamonds$value <- NA #Makes new column
for (i in 1:nrow(diamonds)){
    ifelse(diamonds$price[i] < 5000, diamonds$value[i] <- "Low price", #If below 5000 sets low price
           diamonds$value[i] <- "High Price")} #Else sets high price
```

### Question 10

Sticking with the `diamonds` dataset for this question. A carat is a special unit of weight for gemstones where a single carat is 200 mg or 0.2 g. Write a function to convert the `diamonds$carat` vector into grams or milligrams. How heavy is the heaviest diamond?

```r
data(diamonds) #Loads data set
carat_convert <- function(x){ #Makes function
    x$mass <- NA #Creates mass coloumn
    for (i in 1:nrow(x)){ #Iterates through every row
        x$mass[i] <- x$carat[i]*0.2} #Covnerts carats to g
    return(x)} #returns new df

head(arrange(carat_convert(diamonds), desc(mass)), 1) #Finds heaviest
head(arrange(carat_convert(diamonds), mass), 1) #Finds lightest
#This takes a little bit
```
