### Question 8
Using the diamonds dataset and the plotting code of your choice, plot the following from the diamonds dataset.

1. Carat vs Price in a scatter plot.
2. A histogram of prices.
3. A boxplot of cut vs price.

The code can be loaded from seaborn in the cell below.

````python
import seaborn as sns
diamonds = sns.load_dataset('diamonds')
````

````python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

diamonds = sns.load_dataset('diamonds')

diamonds.plot.scatter(x='carat', y='price', color = "Purple", alpha = 0.25)
plt.title("Scatter graph of the carat of diamonds against their price")
plt.xlabel("Diamond weight (Carat)")
plt.ylabel("Price")
plt.show()

diamonds['price'].plot.hist(bins = 15)
plt.title("Histogram of diamond prices")
plt.xlabel("Price")
plt.show()

diamonds.boxplot(column='price', by='cut', grid=False)
plt.title("Boxplot showing the distribution of diamond price based on their cut")
plt.xlabel("Cut")
plt.ylabel("Price")
plt.show()

````

### Question 10
Write a function that will convert carat to grams and find the heaviest diamond.

1 carat = 0.2 grams

````python
import seaborn as sns
import pandas as pd
diamonds = sns.load_dataset('diamonds')
# ^ Loads dataset and modules
def carat_convert_sort(x): #defines function
    mass = [] #Creates blank list for masses to be added to
    
    for i in range(len(x)): #Loops through the carat values pulled into function
        mass.append(x[i] * 0.2) #converts the ith carat value into grams than appends it to the mass list
        
    return f'The heaviest diamond has a mass of {str(max(mass))}g' #Returns the heaviest diamonds mass in grams

carat_convert_sort(diamonds['carat'])
    
````
